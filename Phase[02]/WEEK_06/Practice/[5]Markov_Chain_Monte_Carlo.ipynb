# Markov Chain Monte Carlo (MCMC)

MCMC is a class of algorithms used to **sample from complex probability distributions** where direct sampling is difficult.

It combines:

- Markov Chains → dependency on previous state
- Monte Carlo → randomness + sampling

---

# Problem Setup

We want to sample from a target distribution:

$$
f(x)
$$

But directly sampling from \(f(x)\) is hard.

So we construct an easier proposal distribution:

$$
g(x)
$$

such that:

$$
g(x) = M f(x)
$$

where:

- \(M\) is a constant ensuring \(g(x)\) dominates \(f(x)\)

---

# Accept-Reject Sampling

This is a basic sampling technique.

## Idea

We sample from \(g(x)\), then decide whether to accept or reject the sample.

---

## Steps

1. Sample \(x \sim g(x)\)
2. Generate \(u \sim \text{Uniform}(0,1)\)
3. Accept \(x\) if:

$$
u \le \frac{f(x)}{g(x)}$$

Otherwise reject and resample.

---

## Intuition

- \(g(x)\) is easy to sample from
- \(f(x)\) is the true distribution we want
- we “filter” samples from \(g(x)\) to match \(f(x)\)

---

# Limitation of Accept-Reject

- inefficient if \(M\) is large
- many rejections in high dimensions
- poor scalability

---

# Markov Chain Monte Carlo (MCMC)

MCMC improves sampling efficiency.

---

## Key Idea

Instead of independent sampling:

> each sample depends on the previous sample

This creates a **Markov Chain**.

---

## Markov Property

$$
x_{t+1} \sim p(x_{t+1} \mid x_t)
$$

So:

- next sample depends only on current sample
- not the full history

---

# Goal of MCMC

Construct a Markov chain whose stationary distribution is:

$$
f(x)
$$

So that after many steps:

- samples behave like they come from \(f(x)\)

---

# Why It Works

Even if the chain starts from a bad distribution:

- it gradually “mixes”
- eventually converges to target distribution

This is called:

## Stationary Distribution

---

# Metropolis-Hastings (Core MCMC Method)

A common MCMC algorithm.

---

## Steps

1. Start with initial value \(x_t\)
2. Propose new sample \(x'\) from proposal distribution \(q(x'|x_t)\)
3. Compute acceptance probability:

$$
A = \min\left(1, \frac{f(x')q(x_t|x')}{f(x_t)q(x'|x_t)}\right)
$$

4. Accept or reject:

- if accepted → \(x_{t+1} = x'\)
- else → \(x_{t+1} = x_t\)

---

# Intuition

- moves randomly through space
- prefers high-probability regions of \(f(x)\)
- still explores low-probability regions occasionally

---

# Comparison

| Method | Idea | Efficiency |
|--------|------|------------|
| Accept-Reject | independent sampling + filtering | low in high dimensions |
| MCMC | dependent sampling (Markov chain) | high efficiency |
| Metropolis-Hastings | guided random walk | widely used |

---

# Key Insight

MCMC works because:

> a carefully designed Markov chain will eventually “forget” its starting point and sample from the true distribution

---

# Summary

- Goal: sample from complex distribution \(f(x)\)
- Accept-Reject: simple but inefficient
- MCMC: uses Markov dependence
- Metropolis-Hastings: most common implementation

MCMC is the backbone of modern Bayesian inference and probabilistic modeling.
