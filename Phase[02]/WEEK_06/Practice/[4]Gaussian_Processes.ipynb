# Gaussian Processes (GP)

A Gaussian Process is a **non-parametric probabilistic model** used for regression and uncertainty estimation.

It defines a distribution over functions instead of fixed parameters.

---

# Gaussian (Normal) Distribution

A Gaussian Process is based on the idea of a **multivariate normal distribution**.

---

# Goal

Given:

- known data points: \(x\)
- observed function values: \(f(x)\)

We want to predict:

- function values at unseen points \(y\)

So essentially:

> learn a distribution over functions and use it to predict unknown outputs

---

# Intuition Example

Imagine:

- a river of length \(L\)
- we observe fish at some known positions
- we want to predict fish density at unseen locations

Instead of fitting a single curve, we model a **distribution of possible curves** that fit the data.

---

# Key Assumption

We assume the function values follow a multivariate Gaussian distribution:

$$
f(X), f(Y) \sim \mathcal{N}(\mu, \Sigma)
$$

---

## Mean Function

Common assumption:

$$
\mu = 0
$$

This simplifies the model:

$$
f(x) \sim \mathcal{GP}(0, k(x,x'))
$$

---

# Covariance Function (Kernel)

The most important part of Gaussian Processes is the covariance function.

It defines how two points are related.

---

## RBF / Squared Exponential Kernel

$$
\Sigma_{i,j}
=
\sigma^2
\exp\left(
-\frac{\|x_i - x_j\|^2}{2l^2}
\right)
$$

---

## Meaning of Parameters

- \( \sigma^2 \): controls vertical variation (output variance)
- \( l \): length scale (smoothness of function)
  - small \(l\) → highly flexible function
  - large \(l\) → smoother function

---

# Joint Distribution

For observed points \(X\) and new points \(Y\):

$$
\begin{bmatrix}
f(X) \\
f(Y)
\end{bmatrix}
\sim
\mathcal{N}
\left(
0,
\begin{bmatrix}
K(X,X) & K(X,Y) \\
K(Y,X) & K(Y,Y)
\end{bmatrix}
\right)
$$

---

# Conditional Distribution

The key idea in GP prediction:

$$
f(Y) \mid f(X)
$$

is also Gaussian.

This gives:

> a full predictive distribution (not just a point estimate)

---

## Interpretation

Instead of predicting a single value, GP gives:

- mean prediction
- uncertainty (variance)

So we get:

- “best guess”
- “confidence in guess”

---

# Why Gaussian Processes?

Gaussian Processes are powerful because:

- work well with small datasets
- provide uncertainty estimation
- flexible (no fixed number of parameters)
- can model complex functions

---

# Summary

## Core Idea

A Gaussian Process defines:

> a distribution over possible functions

---

## Key Components

- Mean function: usually 0
- Kernel (covariance): defines similarity between points

---

## Kernel Example

$$
\Sigma_{i,j}
=
\sigma^2
\exp\left(
-\frac{\|x_i - x_j\|^2}{2l^2}
\right)
$$

---

## Prediction

Given observed data:

- posterior distribution is Gaussian
- gives mean + variance for predictions

---

## Final Insight

Gaussian Processes answer:

> "Instead of learning a function, why not learn a distribution over all possible functions?"
