# Undirected Graphical Models

Undirected Graphical Models represent probability distributions using an **undirected graph** instead of a directed acyclic graph (DAG).

While Bayesian Networks use directed edges to encode causality, undirected models focus on **symmetry and mutual dependence**.

---

# Why Undirected Models?

Not all real-world relationships are naturally directional.

Examples:

- pixels in an image
- spatial data
- physical systems
- constraint-based systems

In these cases, relationships are often **symmetric**, so undirected graphs are more appropriate.

---

# General Formulation

The joint distribution is defined as:

$$
P(X)
=
\frac{1}{Z}
\prod_{c \in \text{cliques}(G)}
\phi_C(x_c)
$$

where:

- \( \phi_C(x_c) \) = potential function over clique \(C\)
- \( Z \) = normalization constant (partition function)
- ensures probabilities sum to 1

---

## Partition Function

$$
Z = \sum_X \prod_{c \in \text{cliques}(G)} \phi_C(x_c)
$$

It normalizes the distribution so that:

$$
\sum_X P(X) = 1
$$

---

# Markov Random Fields (MRFs)

Markov Random Fields are a type of Undirected Graphical Model.

They define a set of conditional independence properties using graph separation.

---

## Key Idea

Any two subsets of variables:

- \(S\)
- \(T\)

are conditionally independent given a separating set.

---

## Conditional Independence Property

If a set \(C\) separates \(S\) and \(T\) in the graph:

$$
S \perp T \mid C
$$

This means:

- once \(C\) is known
- information from \(S\) gives no extra information about \(T\)

---

# Independence Corollaries

MRFs encode independence through **graph separation**:

### 1. Local Markov Property

A node is conditionally independent of all other nodes given its neighbors.

---

### 2. Pairwise Markov Property

Non-adjacent nodes are conditionally independent given all other nodes.

---

### 3. Global Markov Property

Any two sets are independent given a separating set.

---

# Bayesian Networks as MRFs

Although Bayesian Networks are directed and MRFs are undirected, a Bayesian Network can sometimes be converted into an MRF.

### Key Idea:

- Moralization process:
  - remove directions
  - connect parents of common children
  - convert DAG into undirected graph

This allows certain inference techniques to be applied more easily.

---

# Introduction to Belief Propagation

Belief Propagation is an inference algorithm used in graphical models.

It computes marginal probabilities efficiently.

---

## Goal

Compute:

$$
P(X_i)
$$

for each variable \(X_i\), without enumerating the full joint distribution.

---

## Idea

Instead of global computation:

- nodes pass "messages" to neighbors
- each message summarizes local information
- repeated updates refine beliefs

---

## Message Passing Concept

Each node sends messages like:

- “what I believe about you”
- based on my local evidence and neighbors

---

## Types of Belief Propagation

### 1. Sum-Product Algorithm

Used for computing marginals:

$$
P(X_i)
$$

---

### 2. Max-Product Algorithm

Used for most likely configuration:

$$
\arg\max_X P(X)
$$

---

## Why It Works

Belief propagation works efficiently because:

- it exploits conditional independence
- avoids full joint computation
- uses local computations instead of global enumeration

---

# Summary

## Undirected Models

- represent symmetric dependencies
- use clique potentials
- normalized using partition function

---

## Markov Random Fields

- independence defined via graph separation
- no directionality

---

## Belief Propagation

- message passing algorithm
- computes marginals efficiently
- fundamental inference technique in MRFs and factor graphs
