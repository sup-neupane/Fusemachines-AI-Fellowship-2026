# Hidden Markov Models (HMM)

A Hidden Markov Model (HMM) is an extension of a Markov Model where:

- the actual states are hidden (unobserved)
- we only observe outputs generated by those hidden states

Examples:

- Speech Recognition
- Part-of-Speech Tagging
- Weather Prediction
- DNA Sequence Analysis

---

# Structure of an HMM

There are two types of variables:

## Hidden States

$$
x_1, x_2, \dots, x_T
$$

These are not directly observed.

---

## Observations

$$
y_1, y_2, \dots, y_T
$$

These are visible to us.

---

## Graph Structure

```
x1 → x2 → x3 → ... → xT
|    |    |
v    v    v
y1   y2   y3 ... yT
```

---

# Hidden State Transitions

The hidden states follow the Markov property:

$$
p(x_t|x_{t-1})
$$

This is called the **Transition Probability**.

---

# Emission Probability

Each hidden state generates an observation:

$$
p(y_t|x_t)
$$

This is called the **Emission Probability**.

---

# Joint Distribution

The complete joint distribution of hidden states and observations is:

$$
p(X,Y)
=
p(x_1)
\prod_{t=1}^{T-1}
p(x_{t+1}|x_t)
\prod_{t'=1}^{T}
p(y_{t'}|x_{t'})
$$

where:

- \(p(x_1)\) = initial state probability
- \(p(x_{t+1}|x_t)\) = transition probability
- \(p(y_t|x_t)\) = emission probability

---

# Forward Inference

Forward inference computes:

$$
\alpha_t(x_t)
=
p(x_t,y_1,\dots,y_t)
$$

This represents:

- probability of being in state \(x_t\)
- after observing observations up to time \(t\)

---

## Initialization

For \(t=1\):

$$
p(x_1,y_1)
=
p(x_1)p(y_1|x_1)
$$

Thus:

$$
\alpha_1(x_1)
=
p(x_1)p(y_1|x_1)
$$

---

## Second Step

For \(t=2\):

$$
p(x_2,y_1,y_2)
=
\sum_{x_1}
p(x_1,y_1)
p(x_2|x_1)
p(y_2|x_2)
$$

or

$$
\alpha_2(x_2)
=
\sum_{x_1}
\alpha_1(x_1)
p(x_2|x_1)
p(y_2|x_2)
$$

---

## General Forward Recursion

$$
\alpha_{t+1}(x_{t+1})
=
\sum_{x_t}
\alpha_t(x_t)
p(x_{t+1}|x_t)
p(y_{t+1}|x_{t+1})
$$

---

## Intuition

At each step:

1. Take previous forward message
2. Apply transition probabilities
3. Incorporate observation likelihood
4. Produce next forward message

---

# Backward Inference

Backward inference propagates information from the future toward the past.

Define:

$$
\beta_t(x_t)
=
p(y_{t+1},\dots,y_T|x_t)
$$

This represents:

- probability of future observations
- given current hidden state

---

## Initialization

At the final time step:

$$
\beta_T(x_T)
=
1
$$

because there are no future observations left.

---

## Recursion

Starting from:

$$
\beta_{t-1}(x_{t-1})
=
p(y_t,\dots,y_T|x_{t-1})
$$

we obtain:

$$
\beta_{t-1}(x_{t-1})
=
\sum_{x_t}
p(x_t|x_{t-1})
p(y_t,\dots,y_T|x_t)
$$

Using conditional independence:

$$
=
\sum_{x_t}
p(x_t|x_{t-1})
p(y_t|x_t)
p(y_{t+1},\dots,y_T|x_t)
$$

which becomes:

$$
\beta_{t-1}(x_{t-1})
=
\sum_{x_t}
p(x_t|x_{t-1})
p(y_t|x_t)
\beta_t(x_t)
$$

---

# Forward-Backward Algorithm

Forward and Backward messages can be combined.

This process is called:

## Smoothing

or

## Forward-Backward Inference

---

## Posterior Probability

The probability of a hidden state given all observations is:

$$
p(x_t|y_1,\dots,y_T)
\propto
\alpha_t(x_t)\beta_t(x_t)
$$

More precisely:

$$
p(x_t|Y)
=
\frac{
\alpha_t(x_t)\beta_t(x_t)
}
{
p(Y)
}
$$

---

# Learning in HMMs

So far we assumed:

- transition probabilities known
- emission probabilities known

But in practice these parameters are usually unknown.

---

# Baum-Welch Algorithm

Baum-Welch is the standard algorithm used to learn HMM parameters.

It is a special case of:

## Expectation Maximization (EM)

---

## E-Step

Use Forward-Backward algorithm to estimate:

- hidden state probabilities
- transition probabilities

from current model parameters.

---

## M-Step

Update:

- transition matrix
- emission matrix
- initial state probabilities

to maximize likelihood.

---

## Repeat

Repeat E-Step and M-Step until convergence.

---

# Important Problems Solved by HMMs

## Evaluation

Compute:

$$
p(Y)
$$

Probability of observed sequence.

Solved using:

- Forward Algorithm

---

## Decoding

Find most likely hidden state sequence:

$$
X^*
=
\arg\max_X p(X|Y)
$$

Solved using:

- Viterbi Algorithm

---

## Learning

Estimate unknown model parameters.

Solved using:

- Baum-Welch Algorithm

---

# Summary

An HMM consists of:

- Hidden States \(X\)
- Observations \(Y\)
- Transition Probabilities \(p(x_t|x_{t-1})\)
- Emission Probabilities \(p(y_t|x_t)\)

Key Algorithms:

| Task | Algorithm |
|--------|------------|
| Evaluation | Forward Algorithm |
| Smoothing | Forward-Backward |
| Decoding | Viterbi Algorithm |
| Learning | Baum-Welch Algorithm |

HMMs are one of the foundational probabilistic models for sequential and time-series data.
