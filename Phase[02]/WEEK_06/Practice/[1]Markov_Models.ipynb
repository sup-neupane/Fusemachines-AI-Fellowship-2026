# Markov Models

Markov Models are probabilistic graphical models designed for sequential data.

Examples:

- Weather prediction
- Stock prices
- Speech recognition
- Natural Language Processing

---

# Markov Assumption

The fundamental assumption is:

## The past is independent of the future given the present.

In other words:

If we know the current state, knowledge of earlier states provides no additional information about future states.

---

## Mathematical Form

$$
p(x_i, x_k \mid x_j)
=
p(x_i \mid x_j)
p(x_k \mid x_j)
\qquad
i < j < k
$$

This means:

Given the present state:

$$
x_j
$$

the past:

$$
x_i
$$

and future:

$$
x_k
$$

become conditionally independent.

---

## First-Order Markov Property

A state depends only on the previous state:

$$
p(x_{t+1}\mid x_1,\dots,x_t)
=
p(x_{t+1}\mid x_t)
$$

This greatly simplifies modelling.

---

# Joint Distribution

Using the Markov assumption:

$$
p(x_1,\dots,x_T)
=
p(x_1)
\prod_{t=1}^{T-1}
p(x_{t+1}\mid x_t)
$$

where:

- \(p(x_1)\) is the initial state probability
- \(p(x_{t+1}|x_t)\) is the transition probability

---

## Graph Structure

A Markov Chain can be represented as:

```
X1 → X2 → X3 → X4 → ... → XT
```

Each state depends only on the immediately previous state.

---

# Inference in Markov Models

A common task is computing probabilities involving hidden or future states.

Direct computation can be expensive because many variables must be summed out.

To solve this efficiently we use:

- Variable Elimination
- Forward Message Passing

---

# Variable Elimination

Variable Elimination avoids constructing the entire joint distribution.

Instead:

- variables are eliminated one at a time
- intermediate factors are reused
- computations become more efficient

---

## Example

Suppose we want:

$$
P(X_T)
$$

From:

$$
P(X_1,X_2,\dots,X_T)
$$

Instead of expanding the full joint distribution and summing everything at the end, we eliminate variables sequentially.

This reduces computational cost significantly.

---

# Forward Message Passing

Forward Message Passing is a dynamic programming approach for inference in Markov Models.

Instead of repeatedly recomputing probabilities, we pass summarized information forward through the chain.

---

## Forward Message

Define:

$$
\alpha_t(x_t)
=
P(x_t, e_{1:t})
$$

where:

- \(x_t\) = current state
- \(e_{1:t}\) = evidence observed up to time \(t\)

The forward message summarizes all information from the past.

---

## Recursive Update

The forward message is updated as:

$$
\alpha_{t+1}(x_{t+1})
=
P(e_{t+1}\mid x_{t+1})
\sum_{x_t}
P(x_{t+1}\mid x_t)
\alpha_t(x_t)
$$

---

## Intuition

At every step:

1. Take the previous message
2. Apply transition probabilities
3. Incorporate new evidence
4. Produce the next message

Thus information flows forward through the chain.

---

# Advantages of Forward Message Passing

- Efficient inference
- Avoids repeated calculations
- Uses dynamic programming
- Foundation of Hidden Markov Models (HMMs)

---

=
P(X_{t+1}\mid X_t)
$$

---

## Joint Distribution

$$
p(x_1,\dots,x_T)
=
p(x_1)
\prod_{t=1}^{T-1}
p(x_{t+1}\mid x_t)
$$

---

## Inference Methods

### Variable Elimination

- Eliminates variables sequentially
- Reuses computations

### Forward Message Passing

- Passes summarized probabilities forward
- Efficient for long sequences

These ideas form the foundation for more advanced sequence models such as Hidden Markov Models and Dynamic Bayesian Networks.
