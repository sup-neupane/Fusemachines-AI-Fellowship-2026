# Named Entity Recognition (NER)

## Introduction

Named Entity Recognition (NER) is the process of **identifying and classifying named entities** in text into predefined categories such as person, organization, location, date, and company.

NER is an important **Information Extraction** task in Natural Language Processing (NLP).

---

## Example

Sentence:

```
Sundar Pichai is the CEO of Google and lives in California.
```

Recognized entities:

| Entity | Type |
|--------|------|
| Sundar Pichai | Person |
| Google | Organization / Company |
| California | Location |

---

## Common Entity Types

- Person
- Organization / Company
- Location
- Date
- Time
- Money
- Percentage
- Product
- Event

---

## Applications of Named Entity Recognition

### Search Engines

- Improves search by identifying important entities in user queries and documents.

### Recommendation Systems

- Recommends relevant content based on recognized people, organizations, products, or locations.

### Information Extraction

- Extracts structured information from unstructured text.



````python
import spacy

nlp =  spacy.load("en_core_web_sm")
````

````python
nlp.pipe_names
````

````python
doc = nlp("Tesla Inc is going to acquire twitter for $45 billion")

for ent in doc.ents:
  print(ent.text, " | ", ent.label_, " | ", spacy.explain(ent.label_))
````

````python
from spacy import displacy

displacy.render(doc, style="ent")
````

````python
nlp.pipe_labels['ner']
````

````python
doc = nlp("Michael Bloomberg founded Bloomberg in 1982")
for ent in doc.ents:
    print(ent.text, "|", ent.label_, "|", spacy.explain(ent.label_))
````

````python
doc = nlp("Tesla Inc is going to acquire Twitter Inc for $45 billion")
for ent in doc.ents:
    print(ent.text, " | ", ent.label_, " | ", ent.start_char, "|", ent.end_char)
````

Setting custom entities

````python
doc = nlp("Tesla is going to acquire Twitter for $45 billion")
for ent in doc.ents:
    print(ent.text, " | ", ent.label_)
````

````python
s = doc[2:5]
s
````

````python
from spacy.tokens import Span

s1 = Span(doc, 0, 1, label="ORG")
s2 = Span(doc, 5, 6, label="ORG")

doc.set_ents([s1, s2], default="unmodified")
````

````python
for ent in doc.ents:
    print(ent.text, " | ", ent.label_)
````
