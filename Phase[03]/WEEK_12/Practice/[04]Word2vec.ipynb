# Word Embedding

## Introduction

Word Embedding is the process of converting words into **dense numerical vectors** that capture semantic relationships between words.

Example:

Words with similar meanings have similar vector representations.

$$
\text{King} - \text{Man} + \text{Woman} \approx \text{Queen}
$$

---

# Types of Word Representation

## 1. Frequency-Based Methods

- Represent words based on their frequency of occurrence in documents.
- Do not learn semantic relationships deeply.

Examples:

- Bag of Words (BoW)
- TF-IDF

---

## 2. Prediction-Based Methods

- Learn word representations by predicting words based on their surrounding context.
- Neural networks are used to learn embeddings.

Example:

- Word2Vec

---

# Word2Vec

Word2Vec is a **deep learning-based word embedding technique** that learns word vectors by predicting relationships between words.

It can use:

- Pre-trained models
- Self-trained models on custom datasets

The word vector is not the direct objective of training; it is a **by-product of a prediction task**.

---

# Types of Word2Vec

## 1. Continuous Bag of Words (CBOW)

### Idea

CBOW predicts the **target word** using its surrounding context words.

### Example

Sentence:

```
I love machine learning
```

For target word:

```
machine
```

Context words:

```
I, love, learning
```

Training pair:

$$
X = \text{Context Words}
$$

$$
Y = \text{Target Word}
$$

A neural network is trained to predict the target word from context words.

The learned embedding is obtained from the **weights of the hidden layer**.

---

## 2. Skip-Gram

### Idea

Skip-Gram works opposite to CBOW.

It predicts the **context words** using the target word.

Example:

Sentence:

```
I love machine learning
```

Input:

```
machine
```

Output:

```
I, love, learning
```

Training pair:

$$
X = \text{Target Word}
$$

$$
Y = \text{Context Words}
$$

---

# CBOW vs Skip-Gram

| CBOW | Skip-Gram |
|------|-----------|
| Predicts target word from context | Predicts context from target word |
| Faster training | Slower training |
| Works well on small datasets | Works well on large datasets |
| Better for frequent words | Better for rare words |

---

# Advantages of Word Embeddings

## Captures Semantic Meaning

- Similar words have similar vector representations.

## Low Dimensional Representation

- Converts high-dimensional sparse vectors into compact vectors.

## Dense Vectors

- Most vector values contain meaningful information instead of mostly zeros.

````python
import gensim
import os
import numpy as np
import pandas as pd
````

````python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
````

````python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

story = []

data_path = "got_books"

for filename in os.listdir(data_path):

    file_path = os.path.join(data_path, filename)

    if filename.endswith(".txt"):

        try:
            with open(file_path, encoding="utf-8") as f:
                corpus = f.read()

        except UnicodeDecodeError:
            with open(file_path, encoding="latin-1") as f:
                corpus = f.read()

        raw_sent = sent_tokenize(corpus)

        for sent in raw_sent:
            story.append(simple_preprocess(sent))
````

````python
story[:5]
````

````python
model =  gensim.models.Word2Vec(
    window = 10,
    min_count=2
)
````

````python
model.build_vocab(story)
````

````python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
````

````python

model.wv.most_similar('aegon')
````

````python
model.wv.doesnt_match(['jon','rikon','robb','arya','sansa','bran'])
````

````python
model.wv.similarity('jaime','cersei')
````
