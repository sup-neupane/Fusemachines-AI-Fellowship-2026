# Text Representation

## Introduction

Text representation is the process of converting textual data into **numerical vectors** so that machine learning algorithms can process it.

---

# Basic Terms

## Word

- A single token or term in a sentence.

**Example:**

```
Machine
```

---

## Document

- A collection of words forming a sentence, paragraph, or article.

**Example:**

```
Machine learning is powerful.
```

---

## Corpus (Corpora)

- A collection of multiple documents.

**Example:**

```
D1: Machine learning is powerful.
D2: NLP is interesting.
D3: Deep learning is popular.
```

---

## Vocabulary

- The set of all **unique words** present in a corpus.

**Example**

Corpus:

```
D1: I like NLP
D2: I like AI
```

Vocabulary:

```
{I, like, NLP, AI}
```

Vocabulary Size = **4**

---

# One-Hot Encoding

One-Hot Encoding represents each word as a binary vector where **only one position is 1** and all others are 0.

### Example

Vocabulary:

| Word | Index |
|------|------:|
| I | 0 |
| like | 1 |
| NLP | 2 |
| AI | 3 |

Word representations:

| Word | One-Hot Vector |
|------|----------------|
| I | [1, 0, 0, 0] |
| like | [0, 1, 0, 0] |
| NLP | [0, 0, 1, 0] |
| AI | [0, 0, 0, 1] |

### Advantages

- Simple to understand and implement.
- Suitable for small vocabularies.

### Limitations

- Produces high-dimensional sparse vectors.
- Does not capture semantic similarity between words.

---

# Bag of Words (BoW)

Bag of Words represents a document by the **frequency of each vocabulary word**, ignoring grammar and word order.

### Example

Documents:

```
D1: I like NLP
D2: I like AI
```

Vocabulary:

```
[I, like, NLP, AI]
```

Feature Matrix:

| Document | I | like | NLP | AI |
|-----------|--:|-----:|----:|---:|
| D1 | 1 | 1 | 1 | 0 |
| D2 | 1 | 1 | 0 | 1 |

### Advantages

- Simple and effective for text classification.
- Easy to compute.

### Limitations

- Ignores word order.
- Produces sparse vectors.
- Does not capture context or meaning.

---

# N-Grams (Bag of N-Grams)

An **N-gram** is a sequence of **N consecutive words**.

### Types

- **Unigram (N = 1)** → Single word
- **Bigram (N = 2)** → Two consecutive words
- **Trigram (N = 3)** → Three consecutive words

### Example

Sentence:

```
I love machine learning
```

**Unigrams**

```
I, love, machine, learning
```

**Bigrams**

```
I love
love machine
machine learning
```

**Trigrams**

```
I love machine
love machine learning
```

Bag of N-Grams counts the occurrences of these word sequences instead of individual words.

### Advantages

- Captures local word order.
- Preserves some contextual information.

### Limitations

- Feature space grows rapidly with larger N.
- Still cannot capture long-range dependencies.

# TF-IDF (Term Frequency – Inverse Document Frequency)

TF-IDF assigns importance to words based on how frequently they occur in a document and how rare they are across the corpus.

A word receives:
- **High weight** if it appears frequently in one document but rarely in others.
- **Low weight** if it appears in many documents.

---

## Term Frequency (TF)

Measures how often a term appears in a document.

$$
TF(t,d)=\frac{\text{Number of occurrences of }t\text{ in document }d}{\text{Total number of words in document }d}
$$

---

## Inverse Document Frequency (IDF)

Measures how unique a word is across the corpus.

$$
IDF(t)=\log\left(\frac{N}{df(t)}\right)
$$

where:

- $N$ = Total number of documents
- $df(t)$ = Number of documents containing term $t$

---

## TF-IDF Formula

$$
TF\text{-}IDF(t,d)=TF(t,d)\times IDF(t)
$$

---

## Example

Documents:

```
D1: I like NLP
D2: I like AI
```

The words **I** and **like** appear in both documents, so they receive **lower TF-IDF scores**.

The words **NLP** and **AI** appear in only one document each, so they receive **higher TF-IDF scores** because they are more informative.

### Advantages

- Reduces the importance of common words.
- Highlights discriminative words.
- Widely used in information retrieval and document classification.

### Limitations

- Ignores word order and context.
- Cannot capture semantic relationships between words.
