# Part-of-Speech (POS) Tagging

## Introduction

Part-of-Speech (POS) Tagging is the process of **labeling each word in a sentence with its appropriate grammatical category**, such as noun, verb, adjective, or adverb.

POS tagging is an important **text preprocessing** step in many Natural Language Processing (NLP) tasks.

---

## Example

Sentence:

```
The cat sat on the mat.
```

POS Tags:

| Word | POS Tag |
|------|---------|
| The | Determiner (DT) |
| cat | Noun (NN) |
| sat | Verb (VBD) |
| on | Preposition (IN) |
| the | Determiner (DT) |
| mat | Noun (NN) |

---

## Applications of POS Tagging

### Named Entity Recognition (NER)

- Helps identify entities such as names of people, organizations, and locations.

### Question Answering (QA) Systems

- Assists in understanding sentence structure to retrieve accurate answers.

### Chatbots

- Improves language understanding for generating meaningful responses.

### Word Sense Disambiguation (WSD)

- Determines the correct meaning of a word based on its grammatical role and context.

### Text Parsing

- Provides grammatical information required for syntactic parsing.

### Machine Translation

- Improves translation accuracy by identifying the grammatical role of each word.

---

## Advantages

- Improves syntactic and semantic analysis.
- Enhances the performance of many downstream NLP applications.
- Provides grammatical information useful for language understanding.

# POS Tagging Using Hidden Markov Model (HMM)

## How POS Tagging Works

The goal of POS tagging is to assign the **most appropriate Part-of-Speech (POS) tag** to every word in a sentence.

### Example

Sentence:

```
Book the flight.
```

Possible POS tags:

| Word | Possible Tags |
|------|---------------|
| Book | Noun / Verb |
| the | Determiner |
| flight | Noun |

From the context, **Book** is correctly tagged as a **Verb**.

Final tagging:

```
Book/VB  the/DT  flight/NN
```

HMM selects the sequence of tags with the **highest probability**.

---

# Hidden Markov Model (HMM)

A **Hidden Markov Model (HMM)** is a probabilistic model used for POS tagging, where:

- **Observed states** are the words.
- **Hidden states** are the POS tags.

The objective is to find the most probable sequence of hidden POS tags for the observed words.

---

## Steps in HMM POS Tagging

### Step 1: Train the Model

- Train the HMM using a corpus in which every sentence is **manually POS-tagged**.

---

### Step 2: Compute Emission Probabilities

- Count how many times each word appears with each POS tag.
- Compute the probability of a word given a POS tag.

### Emission Probability

$$
P(\text{Word} \mid \text{Tag})
=
\frac{\text{Count(Tag, Word)}}{\text{Count(Tag)}}
$$

Example:

| Tag | Word | Count |
|-----|------|------:|
| Noun | book | 40 |
| Verb | book | 60 |

---

### Step 3: Compute Transition Probabilities

- Count how often one POS tag follows another.
- Compute the probability of transitioning from one tag to the next.

### Transition Probability

$$
P(T_i \mid T_{i-1})
=
\frac{\text{Count}(T_{i-1},T_i)}
{\text{Count}(T_{i-1})}
$$

Example:

| Previous Tag | Current Tag | Count |
|--------------|-------------|------:|
| DT | NN | 120 |
| VB | DT | 80 |
| NN | VB | 50 |

---

### Step 4: Build the Hidden Markov Model

The HMM consists of:

- Hidden states → POS tags
- Observations → Words
- Transition probabilities
- Emission probabilities

---

### Step 5: Predict POS Tags

For a new sentence:

1. Consider possible POS tags for each word.
2. Compute the probability of each possible tag sequence using transition and emission probabilities.
3. Select the tag sequence with the **highest probability**.

---

# Optimization

Checking every possible tag combination is computationally expensive.

Therefore, HMM uses the **Viterbi Algorithm**, a dynamic programming algorithm that efficiently finds the **most probable sequence of POS tags** without evaluating all possible combinations.

---

## Advantages of HMM

- Simple and efficient probabilistic model.
- Considers both word likelihood and tag sequence.
- Works well for sequence prediction problems such as POS tagging.

---

## Limitations

- Assumes each tag depends only on the previous tag (Markov assumption).
- Performance depends on the availability of large labeled datasets.
- Less accurate than modern deep learning models such as BiLSTM and Transformers.
